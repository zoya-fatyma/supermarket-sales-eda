## 2. Data Cleaning & Preparation

The dataset was checked for missing values and duplicate records. The Date column was converted into datetime format, and Month and Day columns were created for time-based analysis.

```python
df["Date"].head()
```

```python
df["Date"] = pd.to_datetime(df["Date"])
```

```python
df["Date"].dtype
```

```python
df["Month"] = df["Date"].dt.month
```

```python
df["Month"].head()
```

```python
df["Day"] = df["Date"].dt.day_name()
```

```python
df["Day"].head()
```

## 3. Exploratory Data Analysis

This section explores sales performance, products, customers, payment methods, ratings, and time-based patterns in the dataset.

```python
df.groupby("Branch")["Total"].sum().sort_values(ascending=False)
```

```python
df.groupby("Product line")["Total"].sum().sort_values(ascending=False)
```

```python
df.groupby("Customer type")["Total"].sum().sort_values(ascending=False)
```

```python
df.groupby("Gender")["Total"].sum().sort_values(ascending=False)
```

## 4. Customer & Product Analysis

```python
gender_sales = df.groupby(["Product line", "Gender"])["Total"].sum().unstack()

gender_sales.plot(kind="bar", figsize=(10, 6))
plt.xlabel("Product line")
plt.ylabel("Total Sales")
plt.title("Total Sales by Product Line and Gender")
plt.xticks(rotation=45)
plt.legend(title="Gender")
plt.show()
```

```python
branch_product = df.groupby(["Branch", "Product line"])["Total"].sum().unstack()

branch_product.plot(kind="bar", figsize=(10, 6))
plt.xlabel("Branch")
plt.ylabel("Total Sales")
plt.title("Total Sales by Branch and Product Line")
plt.xticks(rotation=0)
plt.legend(title="Product line")
plt.show()
```

```python
customer_product = df.groupby(["Customer type", "Product line"]).size().unstack()

customer_product.plot(kind="bar", figsize=(10, 6))
plt.xlabel("Customer Type")
plt.ylabel("Number of Transactions")
plt.title("Customer Type by Product Line")
plt.xticks(rotation=0)
plt.legend(title="Product line")
plt.show()
```

```python
df.groupby("Product line")["Total"].mean().sort_values(ascending=False)
```

## 5. Payment & Customer Behavior

```python
customer_payment = df.groupby(["Customer type", "Payment"]).size().unstack()

customer_payment.plot(kind="bar", figsize=(8, 5))
plt.xlabel("Customer Type")
plt.ylabel("Number of Transactions")
plt.title("Payment Method by Customer Type")
plt.xticks(rotation=0)
plt.legend(title="Payment")
plt.show()
```

```python
payment_sales = df.groupby("Payment")["Total"].sum().sort_values(ascending=False)

payment_sales.plot(kind="bar", figsize=(8, 5))
plt.xlabel("Payment Method")
plt.ylabel("Total Sales")
plt.title("Total Sales by Payment Method")
plt.xticks(rotation=0)
plt.show()
```

```python
df.groupby("Payment")["Total"].mean().sort_values(ascending=False)
```

## 6. Customer Rating Analysis

```python
rating_product = df.groupby("Product line")["Rating"].mean().sort_values(ascending=False)

rating_product.plot(kind="bar", figsize=(10, 6))
plt.xlabel("Product line")
plt.ylabel("Average Rating")
plt.title("Average Customer Rating by Product Line")
plt.xticks(rotation=45)
plt.show()
```

```python
df.groupby("Customer type")["Rating"].mean().sort_values(ascending=False)
```

```python
df.groupby("Gender")["Rating"].mean().sort_values(ascending=False)
```

- Female customers generated higher total sales than male customers.
